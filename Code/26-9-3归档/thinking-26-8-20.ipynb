无论采取哪种格式的剧情脚本，都可以有以下的命令
- lication（background）：更换背景
- music：更换背景音乐
- show_character：更换角色立绘
- expression：更换角色立绘的表情
- next_scene：更换下一个场景或脚本？
- transition：转场效果
- shake：震动效果

对于不需要参数的命令，可以用一个列表来收集

关于角色表情动画，需要有以下几个：
- idle
- talking
- confident-talking
- confident-idle
- laugh-talking

对于东方角色：
- 符卡效果？
- ？

每个角色表情是由帧动画完成的，只需一个循环即可，并且循环内可以复用（倒放）
