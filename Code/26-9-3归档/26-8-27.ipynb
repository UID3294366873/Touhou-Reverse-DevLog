脚本语言设计：
- 后缀名：`.dramex`
- 可以用`__init__.dramex`标记这个文件夹是一个块
    - 块与块可以嵌套
    - 每个块的`__init__.dramex`内要包含
        - 将会用到的角色 及 每个角色可能的不同动画、表情（默认使用第一个）  
          类似如下的结构：
          ```json
          character {
              "灵梦" : [ "idle", "smile", "lahgh"],
              "Phoenix" : ["idle", "smile", "sad"]
          }
          ```
        - 可自定义组合函数（`def`关键字 + `Tab`嵌套）  
          如
          ```python
          def func(time, effects):
              for effect in effects:
                  effect(time)
          ```
          使用函数
          ```python
          from __init__ import func
          func()
          ```
    - 使用 `#` 来添加注释

首先将可能用到的命令分为三类
1. 对话命令  
   如：
   ```text
   灵梦 我是灵梦
   ```
3. 效果命令
   如：
   ```
   shake(time=3)
   ```
   此处参照python函数语法
5. 句内命令
   ```text
   这是$一段${命令 参数1=... 参数2=... }句子
   ```

对话命令语法类似ren'py，为：`角色名 空格 对话文本`，角色名与对话文本将在处理后被放入对话框显示到界面上   
效果命令语法：`func(arguments)`  
`句内命令`用于`对话命令`内，可以用在角色名和对话文本上，例如显示不同颜色，字号，字体，语速等

针对　`句内命令`，又可以分出两类

1. 包裹式：
   针对需要改变某一段范围内文字效果的情况    
   使用 `$ ${命令}`来包裹  
   若需要传入参数，则通过`$ ${命令 参数}`的方式  
   包裹式应当可以嵌套包裹  

2. 插入式    
   针对在播放对话时某一刻需要如震动之类的特效时采用  
   使用 `[]`来包裹命令，其内为具体的命令与参数，如  
   `[shake(time=3)]`
