# DrameX 语法手册

DrameX 剧情文件使用 `.dramex` 扩展名。

---

## 1. 基本规则

### 1.1 编码

文件使用 UTF-8 编码。

### 1.2 空行

空行不会产生任何剧情效果，可以用于分隔段落。

```python
灵梦：早上好。


> 阳光照进了神社。
```

### 1.3 注释

使用 `#` 添加注释：

```python
# 这是注释
灵梦：今天的天气不错。
```

只有行首的 `#` 表示注释。

```
灵梦：今天的话题是 #幻想乡。
```

这里的 `#幻想乡` 会作为普通文本显示。

### 1.4 缩进

选择分支和条件语句使用缩进表示内容范围。

推荐使用两个空格缩进，不要混用 Tab 和空格。

```python
if trust >= 3:
  灵梦：我相信你。
```

---

## 2. 文本

## 2.1 角色对话

格式：

```text
角色名：对话文本
```

示例：

```text
灵梦：你终于来了。
Phoenix：I am ready.
```

也支持半角冒号：

```text
灵梦: 你终于来了。
```

冒号之后的全部内容都属于对话文本。

---

## 2.2 旁白

使用 `>` 表示没有说话人的文本：

```text
> 夜幕降临，神社逐渐安静下来。
```

旁白可以包含句内效果：

```python
> [speed slow]风吹过树林。[/speed]
```

---

## 2.3 对话延续

如果一段对话需要分成多行，可以使用 `|`：

```text
灵梦：这是一段比较长的对话，
| 可以继续写在下一行。
```

多行内容仍然属于同一个角色。

旁白也可以延续：

```text
> 夜色越来越深，
| 远处传来了奇怪的声音。
```

---

## 3. 字符串与参数

### 3.1 不带引号的内容

角色名、标签名、变量名和简单参数通常不需要引号：

```text
@show 灵梦 smile center
@set trust = 1
```

### 3.2 带引号的内容

资源路径和包含空格的文本使用双引号：

```text
@background "locations/hakurei shrine"
@music "bgm/morning theme"
```

### 3.3 转义

字符串中可以使用：

```text
\"    双引号
\\    反斜杠
\n    换行
```

---

## 4. 剧情标签

使用 `::` 声明标签：

```text
:: start

灵梦：故事开始了。

:: forest

灵梦：我们到了森林。
```

标签名在同一个文件内必须唯一。

标签通常用于跳转：

```text
@goto forest
```

---

## 5. 控制命令

控制命令必须以 `@` 开头：

```text
@命令 参数
```

命令名称使用小写字母和下划线。

命令参数不使用括号和逗号：

```text
@shake 0.3
```

不使用：

```text
shake(time=0.3)
```

参数可以分为位置参数和命名参数。

```text
@show 灵梦 smile center
@show 灵梦 smile center duration=0.5
```

位置参数写在前面，命名参数写在后面。

---

## 6. 场景命令

## 6.1 更换背景

```text
@background "locations/shrine"
```

带转场：

```text
@background "locations/forest" transition=fade duration=0.5
```

参数：

```text
@background 资源路径 [transition=转场名称] [duration=持续时间]
```

---

## 6.2 播放音乐

```text
@music "bgm/morning"
```

循环播放：

```text
@music "bgm/battle" loop
```

淡入淡出：

```text
@music "bgm/night" loop fade=1
```

停止音乐：

```text
@music stop
```

---

## 6.3 显示角色

```text
@show 灵梦
```

指定表情：

```text
@show 灵梦 smile
```

指定位置：

```text
@show 灵梦 smile center
```

指定显示时间：

```text
@show 灵梦 smile center duration=0.5
```

常用位置：

```text
left
center
right
```

完整格式：

```text
@show 角色名 [表情] [位置] [duration=持续时间]
```

---

## 6.4 隐藏角色

```text
@hide 灵梦
```

带动画时间：

```text
@hide 灵梦 duration=0.5
```

---

## 6.5 切换表情

```text
@expression 灵梦 angry
```

带动画时间：

```text
@expression 灵梦 sad duration=0.2
```

---

## 6.6 播放转场

```text
@transition fade
```

指定持续时间：

```text
@transition fade duration=0.5
```

---

## 6.7 屏幕震动

```text
@shake 0.3
```

指定强度：

```text
@shake 0.3 strength=0.8
```

格式：

```text
@shake 持续时间 [strength=强度]
```

---

## 6.8 等待

```text
@wait 1
```

表示等待一秒后继续执行。

---

## 6.9 播放音效

```text
@sound "sounds/click"
```

---


## 7. 剧情跳转

## 7.1 当前文件内跳转

```text
@goto forest
```

目标必须是当前文件中的标签。

---

## 7.2 调用其他文件

```text
@call "chapters/forest"
```

也可以调用指定标签：

```text
@call "chapters/forest:entry"
```

被调用的脚本结束后，会返回到原来的位置。

---

## 7.3 无返回跳转

```text
@jump "endings/bad:main"
```

`jump` 不会返回到当前脚本。

---

## 7.4 返回

```text
@return
```

如果脚本由 `@call` 调用，执行到文件末尾时也会自动返回。

---

## 8. 选择分支

使用 `choice:` 声明选择。

### 8.1 简单选择

```text
choice:
  - 去森林看看 -> forest
  - 留在神社 -> shrine
```

格式：

```text
choice:
  - 选项文本 -> 标签
```

选择后会跳转到对应标签。

---

### 8.2 带剧情内容的选择

```text
choice:
  - 去森林看看
    @show 灵梦 serious center
    @call "chapters/forest:entry"
    @goto after_forest

  - 留在神社
    @goto shrine
```

每个选项下面缩进的内容，属于该选项。

---

### 8.3 带条件的选择

```text
choice:
  - 打开宝箱 -> open_chest when has_key
  - 直接离开 -> leave
```

`when` 后面的条件为假时，该选项不会显示。

格式：

```text
- 选项文本 -> 标签 when 条件
```

---

## 9. 条件语句

## 9.1 `if`

```text
if trust >= 3:
  灵梦：我相信你。
```

缩进内容属于条件成立时执行的剧情。

---

## 9.2 `else`

```text
if has_key:
  灵梦：你有钥匙。

else:
  灵梦：你没有钥匙。
```

---

## 9.3 多个条件

```text
if trust >= 3:
  灵梦：我相信你。

else if trust >= 1:
  灵梦：我愿意听你解释。

else:
  灵梦：请你离开。
```

条件块可以包含对话、命令、跳转和选择。

---

## 10. 变量

## 10.1 设置变量

```text
@set trust = 0
@set has_key = true
@set route = "magic"
```

支持的基本类型：

```text
数字
字符串
布尔值
```

布尔值只有：

```text
true
false
```

---

## 10.2 修改数字变量

```text
@set trust += 1
@set trust -= 1
```

也可以直接重新赋值：

```text
@set trust = 5
```

---

## 10.3 删除变量

```text
@unset has_key
```

---

## 10.4 条件中使用变量

```text
if trust >= 3:
  灵梦：看来我们已经是朋友了。
```

```text
if has_key:
  > 你拿出了钥匙。
```

---

## 10.5 变量插值

在文本中使用 `{变量名}`：

```text
灵梦：你好，{player_name}。
```

```text
> 你的好感度目前是 {trust}。
```

变量不存在时，编辑器应当提示错误。

---


## 11. 条件表达式

支持以下形式：

```text
trust >= 3
trust == 5
route == "magic"
has_key
not has_key
```

比较运算符：

```text
==    等于
!=    不等于
>     大于
>=    大于等于
<     小于
<=    小于等于
```

逻辑运算符：

```text
and
or
not
```

示例：

```text
if trust >= 3 and has_key:
  灵梦：你已经获得了进入地下室的资格。
```

```text
if route == "magic" or route == "sword":
  > 你选择了一条战斗路线。
```

也支持括号：

```text
if (trust >= 3 and has_key) or route == "magic":
  > 条件成立。
```

## 12. 句内效果

句内效果使用方括号。

---

## 12.1 包裹式效果

```text
灵梦：这是一段 [wave]会摆动的文字[/wave]。
```

基本格式：

```text
[效果名 参数]文字[/效果名]
```

示例：

```text
灵梦：[color red]危险[/color]！
```

```text
灵梦：[speed slow]这句话会慢慢显示。[/speed]
```

---

## 12.2 插入式效果

一次性效果使用自闭合标签：

```text
灵梦：等等，[pause 0.3/]你听见了吗？
```

格式：

```text
[效果名 参数/]
```

示例：

```text
灵梦：就在这时，[shake 0.5/]房间突然震动起来。
```

```text
灵梦：请听我说，[sound "click"/]这很重要。
```

---

## 12.3 常用句内效果

```text
[wave]文字[/wave]
[color red]文字[/color]
[speed slow]文字[/speed]
[pause 0.3/]
[shake 0.5/]
[sound "click"/]
```

句内效果的参数规则与控制命令一致：

```text
[color red]文字[/color]
```

也可以写成：

```text
[color value=red]文字[/color]
```

---

## 12.4 嵌套效果

句内效果可以嵌套：

```text
灵梦：[wave][color red]危险[/color][/wave]！
```

标签必须正确闭合。

正确：

```text
[wave][color red]文字[/color][/wave]
```

错误：

```text
[wave][color red]文字[/wave][/color]
```

句内标签不能跨越多行。

---

## 12.5 显示特殊符号

如果需要显示方括号或花括号，可以使用反斜杠：

```text
灵梦：这是 \[普通文本\]。
```

```text
灵梦：变量写法是 \{name\}。
```

## 13. 角色声明

角色声明用于记录角色和表情名称。

```text
character 灵梦 idle smile laugh
character Phoenix idle smile sad
```

格式：

```text
character 角色名 表情名...
```

角色名包含空格时使用双引号：

```text
character "神秘声音" idle angry
```

角色声明通常写在文件开头：

```text
character 灵梦 idle smile laugh

:: start

@show 灵梦 smile center
灵梦：你好。
```

## 14. 自定义命令与命令组合

DrameX 支持使用 `@define` 定义由多个命令组成的自定义命令。

---

### 14.1 定义命令组合

使用 `@define` 定义命令组合：

```text
@define dramatic_pause:
  @wait 0.3
  @shake 0.2
```

使用时直接调用命令名：

```text
@dramatic_pause
```

执行效果等同于：

```text
@wait 0.3
@shake 0.2
```

命令组合中的命令会按照书写顺序依次执行。

---

### 14.2 带参数的自定义命令

定义参数：

```text
@define dramatic_pause duration=0.3 strength=0.8:
  @wait {duration}
  @shake {duration} strength={strength}
```

调用时可以使用默认参数：

```text
@dramatic_pause
```

也可以覆盖默认参数：

```text
@dramatic_pause duration=0.6 strength=1
```

也支持位置参数：

```text
@dramatic_pause 0.6 1
```

参数按照定义时的顺序匹配。

---

### 14.3 必填参数

不带默认值的参数为必填参数：

```text
@define change_background background duration=0.5:
  @background {background} transition=fade duration={duration}
```

调用：

```text
@change_background "locations/forest"
```

```text
@change_background "locations/forest" duration=1
```

如果缺少 `background` 参数，编辑器应当提示错误。

---

### 14.4 参数引用

在自定义命令内部，使用 `{参数名}` 引用参数：

```text
@define show_character character expression position:
  @show {character} {expression} {position}
```

调用：

```text
@show_character 灵梦 smile center
```

参数引用只对自定义命令内部的命令参数生效。

---

### 14.5 自定义命令嵌套

自定义命令可以调用其他自定义命令：

```text
@define dramatic_pause:
  @wait 0.3
  @shake 0.2

@define dramatic_entry:
  @transition fade duration=0.5
  @show 灵梦 serious center
  @dramatic_pause
```

调用：

```text
@dramatic_entry
```

执行顺序为：

```text
@transition fade duration=0.5
@show 灵梦 serious center
@wait 0.3
@shake 0.2
```

---

### 14.6 自定义命令的作用范围

自定义命令在当前 `.dramex` 文件中有效。

建议将自定义命令写在文件开头，并放在第一个剧情标签之前：

```text
@define dramatic_pause:
  @wait 0.3
  @shake 0.2

:: start

灵梦：故事开始了。
@dramatic_pause
```

---

### 14.7 自定义命令的限制

自定义命令组合中可以使用：

```text
@background
@music
@show
@hide
@expression
@transition
@shake
@wait
@sound
@set
其他自定义命令
```

第一版中，自定义命令组合不应包含：

```text
:: 标签
choice:
if
@goto
@jump
@call
@return
```

自定义命令主要用于组合演出命令，不用于创建新的剧情流程。

---

### 14.8 命名规则

自定义命令名称不能与内置命令重名。
