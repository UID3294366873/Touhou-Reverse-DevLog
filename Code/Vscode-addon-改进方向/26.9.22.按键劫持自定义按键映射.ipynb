后续改进方向：

由于主要是中文的对话文本，会导致在写脚本的时候反复多次切换中英文输入法\
为了解决这个问题，首先使用快捷键`ctrl+.`可以设置中文输入法时使用英文标点符号\
但是这样一来在写对话文本时又出现了无法输出`……``，``。`等问题\
为了解决该问题，或许我可以在插件中写入操控键盘映射的逻辑\
例如
当使用中文输入法并且使用了`ctrl+.`切换标点符号的前提下\
- 按下`^`键，会输出`……`
- 按下`.`键，会输出`。`
- 按下`,`键，会输出`，`

---
需要注意的是，如采用了这种方案，则要输出`<`或是`>`成了新的问题\
也许我可以用
- `ctrl + ,`来输出`<`
- `ctrl + .`来输出`>`

---

哦其实`<`和`>`是`shift+,`的输出，其实不影响

若vscode内置的快捷键不常用，可以考虑在启动插件后对其直接禁用
