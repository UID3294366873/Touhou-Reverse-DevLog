是预先解析完毕脚本，存储为json或是godot的resource  
还是按需解析，在游戏运行到哪部分脚本再解析哪部分脚本？

目前倾向于预先解析

解析脚本的时候，需要一个字典，键1是脚本文件名（无后缀），值1是脚本资源路径，键2是字符串“anchors”，用于存储键1脚本内所有的剧情标签，值2则是一个字典，该字典中键值对分别为剧情标签字符串以及其在该脚本中对应的行号（空间换时间），格式如下：
```python
{
    <DraemeX>: <Path>,
    "anchors": {
        <anchor1>: <idx>,
        <anchor2>: <idx>,
        ...
    }
}
```

但是这样只能用于存储一个dramex文件，或许我可以整一个json文件，用于存储这些内容，格式如下：

```json
[
    {
        <DraemeX>: <Path>,
        "anchors": {
            <anchor1>: <idx>,
            <anchor2>: <idx>,
            ...
        }
    },
    ...
]

用作列表的话检索是`O(n)`，我得改写成字典
```json
{
    <Dramex_file_name> : {
        "path": <Dramex_file_path>,
        "anchors": {
            <anchor1>: <idx>,
            <anchor2>: <idx>,
            ...
        }
    },

    ...
}

最后将这个json文件存储为`index.json`\
这个json的作用作为剧情脚本翻译结果散列表的索引，以`O(1)`的速度，快速进行剧情跳转\
该json文件将在解析`.dramex`剧情脚本的时候一并进行解析
